#**Business Understanding**
##**Project overview**
This project uses the World Health Organization (WHO) Global Health Observatory dataset to examine trends in the estimated number of people living with HIV across countries from 2000 to 2023.
The analysis focuses on two questions:
1. *Which countries collectively account for approximately 75% of the global reported HIV burden, and how has their burden changed over time?*
2. *Which countries within the WHO African Region collectively account for approximately 75% of the region's reported HIV burden, and what trends do they exhibit?*

The analysis will use data preparation, cleaning, exploratory data analysis (EDA), cumulative burden calculations, and time-series visualizations to produce insights useful for public health planning and resource allocation.

**Important distinction**: The dataset measures the estimated number of people living with HIV, not the number of new HIV infections. These figures reflect the existing burden of HIV, including people who may be receiving treatment.

##**Objective**
###**Primary objective**
To analyze and visualize the distribution and trends of people living with HIV in countries that collectively account for approximately 75% of the reported global burden and the reported burden within the WHO African Region.

####**Specific objectives**
1. Identify high-burden countries: Rank countries by their estimated number of people living with HIV in 2023 and select countries accounting for approximately 75% of the reported total.
2. Analyze global trends: Examine how the estimated number of people living with HIV has changed in the selected countries between 2000 and 2023.
3. Analyze African regional trends: Identify the countries accounting for approximately 75% of the reported African regional burden and compare their trends.
4. Identify public health priorities: Highlight countries with persistently high or increasing numbers of people living with HIV.
5. Support evidence-based decisions: Translate the findings into insights that can inform HIV prevention, treatment, testing, and healthcare resource planning.

#**Business Value**
This analysis provides actionable insights into the distribution and long-term trends of the HIV burden within the WHO African Region. By identifying countries that collectively account for at least 75% of people living with HIV, the analysis supports evidence-based resource allocation, targeted intervention planning, policy development, and monitoring of public health progress. The findings can help governments, healthcare providers, and funding organizations prioritize services and plan for treatment needs, ultimately supporting more efficient use of limited healthcare resources and improved HIV program outcomes.

**Expected value**: Better targeting of limited public health resources and clearer monitoring of HIV burden over time. The analysis can inform decisions, but it does not independently establish which interventions caused changes in HIV trends.
